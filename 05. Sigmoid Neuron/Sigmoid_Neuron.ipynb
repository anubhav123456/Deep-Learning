# Importing Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d
import matplotlib.colors as mcolors
import seaborn as sns
sns.set_theme(style="darkgrid")
```

# $$f(wx+b) = \frac{1}{1 + e^{-(wx + b)}}$$

```python
def sigmoid(x, w, b):
  temp1 = w * x + b;
  temp2 = 1 + np.exp(-temp1);
  return 1/temp2;
```

```python
print(sigmoid(1, 0.5, 0));
```

* **Positive $w$:** The sigmoid curve increases from 0 to 1.

```python
w = 0.9;
b = 1;
X = np.linspace(-10, 10, 100);
Y = sigmoid(X, w, b);
plt.plot(X, Y);
plt.show();
```

* **Negative $w$:** The sigmoid curve decreases from 1 to 0.

```python
w = -0.9;
b = 1;
X = np.linspace(-10, 10, 100);
Y = sigmoid(X, w, b);
plt.plot(X, Y);
plt.show();
```

* **Small magnitude of $w$** (e.g., $w = 0.3$) $\rightarrow$ Gradual, smooth transition.

```python
w = 0.9;
b = 1;
X = np.linspace(-10, 10, 100);
Y = sigmoid(X, w, b);
plt.plot(X, Y);
plt.show();
```

* **Large magnitude of $w$** (e.g., $w = 3$, $4$, or $16$) $\rightarrow$ Very steep transition.

```python
w = 16;
b = 1;
X = np.linspace(-10, 10, 100);
Y = sigmoid(X, w, b);
plt.plot(X, Y);
plt.show();
```

* **Increasing $b$** (assuming $w > 0$): Shifts the curve to the **left**.

```python
w = 0.9;
b = 1;
X = np.linspace(-10, 10, 100);
Y = sigmoid(X, w, b);
plt.plot(X, Y);
plt.show();
```

```python
w = 0.9;
b = 4;
X = np.linspace(-10, 10, 100);
Y = sigmoid(X, w, b);
plt.plot(X, Y);
plt.show();
```

* **Decreasing $b$** (making $b$ more negative, assuming $w > 0$): Shifts the curve to the **right**.

```python
w = 0.9;
b = -4;
X = np.linspace(-10, 10, 100);
Y = sigmoid(X, w, b);
plt.plot(X, Y);
plt.show();
```

* **Decreasing $b$** (making $b$ more negative, assuming $w < 0$): Shifts the curve to the **left**.

```python
w = -0.9;
b = -4;
X = np.linspace(-10, 10, 100);
Y = sigmoid(X, w, b);
plt.plot(X, Y);
plt.show();
```

* **Increasing $b$** (making $b$ more positive, assuming $w < 0$): Shifts the curve to the **right**.

```python
w = -0.9;
b = 4;
X = np.linspace(-10, 10, 100);
Y = sigmoid(X, w, b);
plt.plot(X, Y);
plt.show();
```

# $$S_{w_1, w_2, b}(x_1, x_2) = \frac{1}{1 + e^{-(w_1 x_1 + w_2 x_2 + b)}}$$

```python
def sigmoid_2d(x1, x2, w1, w2, b):
  return 1 / (1 + np.exp(-(w1 * x1 + w2 * x2 + b)))
```

```python
# Example function call
sigmoid_2d(1, 0, 0.5, 0, 0);
```

# Plotting Loss

```python
w_unknown = 0.5
b_unknown = 0.25

X = np.random.random(25) * 20 - 10
Y = sigmoid(X, w_unknown, b_unknown)
```

```python
plt.plot(X, Y, '*')
plt.show()
```

```python
def calculate_loss(X, Y, w_est, b_est):
  loss = 0
  for x, y in zip(X, Y):
    loss += (y - sigmoid(x, w_est, b_est)) ** 2  # squared loss calculation
  return loss
```

```python
W = np.linspace(0, 2, 100)
B = np.linspace(-1, 1, 100)

WW, BB = np.meshgrid(W, B)

Loss = np.zeros(WW.shape)
```

```python
WW.shape # (100, 100)
```

```python
for i in range(WW.shape[0]):
  for j in range(WW.shape[1]):
    Loss[i, j] = calculate_loss(X, Y, WW[i, j], BB[i, j])
```

```python
fig = plt.figure()
ax = plt.axes(projection='3d')
ax.plot_surface(WW, BB, Loss, cmap='viridis')
ax.set_xlabel('W')
ax.set_ylabel('B')
ax.set_zlabel('Loss')

# ax.view_init(30, 100)
plt.show()
```

```python
fig = plt.figure()
ax = plt.axes(projection='3d')
ax.plot_surface(WW, BB, Loss, cmap='viridis')
ax.set_xlabel('W')
ax.set_ylabel('B')
ax.set_zlabel('Loss')
ax.view_init(30, 180);
```

```python
fig = plt.figure()
ax = plt.axes(projection='3d')
ax.plot_surface(WW, BB, Loss, cmap='viridis')
ax.set_xlabel('W')
ax.set_ylabel('B')
ax.set_zlabel('Loss')
ax.view_init(30, 270);
```

```python
ij = np.argmin(Loss) # returns index of the min value index of the flattened array
```

```python
i = np.floor(ij/Loss.shape[1])
j = ij - i * Loss.shape[1]
i, j = int(i), int(j)
```
